```python
import pandas as pd

messages = pd.read_csv(
    'SMSSpamCollection.txt',
    sep='\t',
    header=None,
    names=['label', 'message']
)

messages
```

```python
##data cleaning and preprocessing
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps = PorterStemmer()
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
lemmatizer = WordNetLemmatizer()
```

```python
corpus = []
for i in range(0, len(messages)):
    # Remove all characters except alphabets and keep spaces
    review = re.sub('[^a-zA-Z ]', ' ', messages['message'][i])
    
    review = review.lower()
    review = review.split()
    # Remove stopwords and apply stemming
    review = [lemmatizer.lemmatize(word , pos='v') for word in review if word not in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer(max_features=2500 , binary = True ,ngram_range=(1, 2) )
```

```python
## INDEPEMDENT feature
x  = vectorizer.fit_transform(corpus).toarray()
```

```python
vectorizer.vocabulary_
```

```python
y = pd.get_dummies(messages['label'] ,  dtype=int)
```

```python
y
```

```python
y = y.iloc[:,0].values
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train, y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_detect_model = MultinomialNB()
```

```python
spam_detect_model.fit(x_train , y_train)
```

```python
y_pred = spam_detect_model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score , classification_report
```

```python
accuracy_score(y_test , y_pred)
```

```python
print(classification_report(y_test , y_pred))
```

```python
msg = ["Win a FREE iPhone now!"]
msg_vec = vectorizer.transform(msg)   # vectorize message
y_pred = spam_detect_model.predict(msg_vec)

if y_pred[0] == 0:
    print("Message is SPAM")
else:
    print("Message is NOT spam")
```
